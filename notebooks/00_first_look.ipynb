```python
import pandas as pd
df = pd.read_csv("../data/raw/online_retail_II.csv",
                 dtype={"Invoice": str, "StockCode": str},
                 parse_dates=["InvoiceDate"])
print(df.shape)
df.head()
```

```python
df.dtypes
```

```python
df.isna().sum()
```

```python
print("Date range:", df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())
print("Cancellation lines:", df["Invoice"].str.startswith("C").sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Price <= 0:", (df["Price"] <= 0).sum())
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
print("Customers with >1 country:",
      (df.groupby("Customer ID")["Country"].nunique() > 1).sum())
print("Stock codes with >1 description:",
      (df.groupby("StockCode")["Description"].nunique() > 1).sum())
print("Invoices with >1 customer:",
      (df.groupby("Invoice")["Customer ID"].nunique() > 1).sum())
print("Invoices with >1 timestamp:",
      (df.groupby("Invoice")["InvoiceDate"].nunique() > 1).sum())
```
